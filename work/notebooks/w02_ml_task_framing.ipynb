## 1. My lane as an ML task (type)

This lane is a scoring / ranking task. We are not classifying a single yes/no outcome; we are assigning a priority score to each content item so an editor can review the most urgent items first. The decision is: `Which refreshed content pieces deserve attention now?` That makes this a ranking problem, with a score that can be sorted by urgency and reviewed in limited batches.

```python
task_type = 'scoring / ranking'
decision = 'Which refreshed content items should editors review first?'
print(f'Task type: {task_type}')
print(f'Decision: {decision}')
```

## 2. Target or proxy

The starter proxy target is `trend_direction == "down"`. In other words, we label a content item as declining when its recent trend direction is down. This is a practical proxy because it is observed in the current data and directly reflects a deterioration in content momentum. In a future warehouse-based model, we would replace this proxy with an observed future-window outcome, such as whether impressions or engagement decline over a later period (for example, the next 30 or 90 days), so the label is tied to a real outcome rather than a contemporaneous directional heuristic.

```python
proxy_target = "trend_direction == 'down'"
future_window_target = 'future-window decline in impressions or engagement'
print(f'Starter proxy: {proxy_target}')
print(f'Future warehouse label: {future_window_target}')
```

## 3. Success metric

We should optimize for Precision@K, because the editorial team can only review a small number of content items each day or week. For example, Precision@50 measures how many of the top 50 scored items are actually relevant decline opportunities. This matches the decision: when a reviewer has capacity for 50 items, the model should put the highest-value opportunities at the top of the queue.

```python
k = 50
relevant_items_in_top_k = 'count of true decline items in the top 50'
print(f'Precision@{k} = {relevant_items_in_top_k} / {k}')
```

## 4. The unit of analysis, as a real dataframe

This dataset is at the level of a single pseudonymized content item. One row = one content item, with trailing-90-day performance and trend metadata. The model learns from per-item signals such as impressions, CTR, engagement, scroll rate, and age to estimate which items are most likely to be declining.

```python
from pathlib import Path

import pandas as pd

# Try likely repo layouts so the notebook works from either the repo root or work/notebooks
paths = [
    Path.cwd() / 'data' / 'raw' / 'content_refresh_anonymized.csv',
    Path.cwd().parent / 'data' / 'raw' / 'content_refresh_anonymized.csv',
    Path('/workspaces/ml-starter-pack-flyrank-ai') / 'data' / 'raw' / 'content_refresh_anonymized.csv',
]

data_path = next((p for p in paths if p.exists()), paths[0])

df = pd.read_csv(data_path)
df['target_decline'] = (df['trend_direction'] == 'down').astype(bool)

print(df[['content_id', 'impressions_90d', 'trend_direction', 'target_decline']].head())
```

## 5. Why ML beats a fixed rule here

A hard rule such as `content_age_days > 180` is too rigid because decline is driven by multiple interacting signals, not one threshold. A learned model can weigh several continuous signals together — CTR, engagement rate, scroll rate, trend direction, impression volume, and age — to capture nuanced cases where older content is not always the problem and newer content can still be underperforming. This lets the model rank items by urgency rather than rely on a brittle boolean cutoff.

```python
signals = ['ctr', 'engagement_rate', 'scroll_rate', 'content_age_days', 'impressions_90d']
print('A learned model weighs multiple continuous signals:', ', '.join(signals))
```
